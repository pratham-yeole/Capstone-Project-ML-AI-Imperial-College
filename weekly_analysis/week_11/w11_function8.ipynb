## **Bayesian Optimisation** - Gaussian Process Regressor

- **Model:** Matern (nu=2.5) + WhiteKernel with Yeo-Johnson transformation. n_restarts_optimizer=25 maintained.

```python
Y_transformed = power_transform(Y_updated.reshape(-1, 1), method='yeo-johnson').ravel()

print('Original Y:', Y_updated)
print('Transformed Y:', Y_transformed)

kernel = Matern(nu=2.5) + WhiteKernel(noise_level=1e-5)

model = GaussianProcessRegressor(
    kernel=kernel,
    n_restarts_optimizer=20)

model.fit(X_unique, Y_transformed)
```

## **Acquisition Function**

- **Key change:** Centring on a blended coordinate, the w10 result [0.153, 0.374, 0.010, 0.236, 0.756, 0.450, 0.195, 0.616] for dimensions 1–4 and 6–8, but using x5 = 0.990 (biased toward the W3 best). 
    - This hybrid approach avoids over-fixating on either the old w3 best or the new w10 result by combining the best-performing values from each.

- **Limitation:** The stochastic nature means any individual result could be noise. 
    - Three consecutive improvements are encouraging but not conclusive. 
    - The 30% uniform component is retained to prevent the 8D search from converging prematurely on what may be a noise-inflated local maximum.

```python
def upper_confidence_bound(X, model, beta=2.0):
    mu, sigma = model.predict(X, return_std=True)
    return mu + beta * sigma

# Blended centre: w10 dims 1-4,6-8 but x5 biased toward w3 best (0.990)
w_blend = np.array([0.153344, 0.374338, 0.010438, 0.235838, 0.99, 0.450377, 0.195188, 0.615623])
x_focused = np.random.normal(loc=w_blend, scale=0.02, size=(7000, 8))
x_uniform = np.random.uniform(0, 1, size=(3000, 8))
x_grid = np.clip(np.vstack([x_focused, x_uniform]), 0, 1)

ucb_values = upper_confidence_bound(x_grid, model, beta=2.0)
next_query = x_grid[np.argmax(ucb_values)]

print(f'Strategic Week 11 Query (Function 8): {next_query[0]:.6f}-{next_query[1]:.6f}-{next_query[2]:.6f}-{next_query[3]:.6f}-{next_query[4]:.6f}-{next_query[5]:.6f}-{next_query[6]:.6f}-{next_query[7]:.6f}')
```
