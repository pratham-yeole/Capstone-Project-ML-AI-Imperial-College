## **Bayesian Optimisation** - Gaussian Process Regressor

- **Model:** Matern (nu=2.5) + WhiteKernel + Yeo-Johnson.

- **On Neural Networks:** Function 4 is the most challenging candidate for a NN surrogate due to its rugged, multi-modal landscape. 
    - However, with ~35 data points in 4D, any NN would overfit severely. 
    - The GP's smooth interpolation with uncertainty bounds is still more appropriate here.

```python
Y_transformed = power_transform(Y_updated.reshape(-1, 1), method='yeo-johnson').ravel()

print('Original Y:', Y_updated)
print('Transformed Y:', Y_transformed)

kernel = Matern(nu=2.5) + WhiteKernel(noise_level=1e-5)

model = GaussianProcessRegressor(
    kernel=kernel,
    n_restarts_optimizer=20)

model.fit(X_unique, Y_transformed)
```

## **Acquisition Function**

- **Key change:** Revert split to 80/20 (from 70/30). 
    - The 70/30 worked as a recovery mechanism but now that we're back on the slope, tighter exploitation is warranted. 

    - Re-centre on the w6 best [0.403, 0.361, 0.354, 0.347].

```python
def expected_improvement(X, model, y_max, xi=0.05):
    mu, sigma = model.predict(X, return_std=True)
    sigma = sigma.reshape(-1, 1); mu = mu.reshape(-1, 1)

    with np.errstate(divide='warn'):
        improvement = mu - y_max - xi
        Z = improvement / sigma
        ei = improvement * norm.cdf(Z) + sigma * norm.pdf(Z)
        ei[sigma == 0.0] = 0.0
    return ei

w6_best = np.array([0.402567, 0.361014, 0.353681, 0.347064])
x_grid = np.random.normal(loc=w6_best, scale=0.015, size=(8000, 4))
x_uniform = np.random.uniform(0, 1, size=(2000, 4))
x_grid = np.vstack([x_grid, x_uniform])
x_grid = np.clip(x_grid, 0, 1)

y_max = np.max(Y_transformed)
ei_values = expected_improvement(x_grid, model, y_max, xi=0.05)
next_query = x_grid[np.argmax(ei_values)]

print(f'Best coordinates: {w6_best}')
print(f'Strategic Week 12 Query (Function 4): {next_query[0]:.6f}-{next_query[1]:.6f}-{next_query[2]:.6f}-{next_query[3]:.6f}')
```
