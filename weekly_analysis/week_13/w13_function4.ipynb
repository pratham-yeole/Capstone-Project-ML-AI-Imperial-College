## **Bayesian Optimisation** - Gaussian Processes

- The Matérn (ν=2.5) + WhiteKernel GP is retained. 
    - For a warehouse logistics function described as having many local optima and sharp efficiency cliffs, the choice of ν=2.5 over the rougher ν=1.5 may seem counterintuitive. 

- However, the sharp output swings seen in this function are not caused by a discontinuous underlying surface as they are caused by a very narrow global peak surrounded by a rapidly declining slope. 
    - A smooth ν=2.5 kernel interpolates correctly between the few positive results and the many negative ones, producing a surrogate that peaks sharply near [0.35–0.41] and falls steeply elsewhere. 
    - A rougher ν=1.5 kernel would add local variation that the data does not support, potentially misleading the acquisition function. 
    
- The Yeo-Johnson transformation is essential here: the output range from -21.857 (w1 random) to +0.602 (w6 best) is wider than any other function in this project, and without transformation the GP's internal scaling would be dominated by the extreme negative values.

- The three consecutive positive results clustering around [0.35–0.40] for all four dimensions provide the GP with its clearest pattern yet for this function. 
    - The remaining challenge is that the w6 best coordinates [0.403, 0.361, 0.354, 0.347] have x1=0.403 are consistently higher than recent queries. 

- This week's centring on the W6 best explicitly addresses this: the GP receives the correct anchor for all four dimensions, and the EI surface will reflect the gradient information accumulated across twelve weeks of positive and negative evidence.

```python
Y_transformed = power_transform(Y_updated.reshape(-1, 1), method='yeo-johnson').ravel()

print('Original Y:', Y_updated)
print('Transformed Y:', Y_transformed)

kernel = Matern(nu=2.5) + WhiteKernel(noise_level=1e-5)

model = GaussianProcessRegressor(
    kernel=kernel,
    n_restarts_optimizer=20)

model.fit(X_unique, Y_transformed)
```

## **Acquisition Function**

- EI with ξ=0.05 and an 80/20 focused-to-uniform split is maintained. 
    - The ξ=0.05 value is deliberately higher than used for other functions as this is intentional. 
    
- The rugged landscape of F4 means the GP's prediction at any given point carries significant uncertainty. 

    - A low ξ would make EI overly conservative, requiring extremely high confidence before suggesting a query that could beat 0.602. 

    - A moderate ξ=0.05 tells EI to accept points with a reasonable probability of improvement even if the prediction is uncertain, which is appropriate when the surrogate itself has limited confidence in the landscape shape. 
    
- The 80/20 split reflects the balance between the confirmed positive zone (exploit) and the need to prevent the search from recursively tightening around a near-miss region. 

- With std=0.015, the focused component keeps 95% of samples within [0.373, 0.433] for x1, directly targeting the gap between the current cluster mean (~0.38) and the W6 best (0.403).

```python
def expected_improvement(X, model, y_max, xi=0.05):
    mu, sigma = model.predict(X, return_std=True)
    sigma = sigma.reshape(-1, 1); mu = mu.reshape(-1, 1)

    with np.errstate(divide='warn'):
        improvement = mu - y_max - xi
        Z = improvement / sigma
        ei = improvement * norm.cdf(Z) + sigma * norm.pdf(Z)
        ei[sigma == 0.0] = 0.0
    return ei

w_best = np.array([0.402567, 0.361014, 0.353681, 0.347064])

x_focused = np.random.normal(loc=w_best, scale=0.015, size=(8000, 4))
x_uniform = np.random.uniform(0, 1, size=(2000, 4))
x_grid = np.vstack([x_focused, x_uniform])
x_grid = np.clip(x_grid, 0, 1)

y_max = np.max(Y_transformed)
ei_values = expected_improvement(x_grid, model, y_max, xi=0.05)
next_query = x_grid[np.argmax(ei_values)]

print(f'Best coordinates: {w_best}')
print(f'Week 13 Query Recommendation (Function 4): {next_query[0]:.6f}-{next_query[1]:.6f}-{next_query[2]:.6f}-{next_query[3]:.6f}')
```
