## **Bayesian Optimisation**

- The Matern (ν=2.5) + WhiteKernel GP is retained. 
    - For an ML model performance surface, which is continuous but can have multiple local maxima corresponding to different viable hyperparameter configurations
        - The ν=2.5 strikes the right balance. 
        
- The WhiteKernel is particularly important here because ML evaluation scores carry inherent stochasticity from random weight initialisation, data shuffling, and numerical precision, all of which introduce evaluation noise that must be absorbed rather than fitted.

- The w12 crash from 2.614 to 0.166 is diagnostic: the query had x3=0.450 (at the ceiling), x1=0.598 (far above the optimal ~0.178), and x6=0.997 (at the boundary). 
    - This combination produced essentially the worst result in the function's recent history. 
    
    - The GP now has a very strong negative gradient signal from this observation, it knows that the region around [0.598, 0.013, 0.450, 0.600, 0.099, 0.997] is catastrophically bad, which sharpens the EI surface around the W11 best [0.178, 0.332, 0.426, 0.199, 0.304, 0.745] considerably.

- The x3 ceiling is lowered from 0.450 to 0.440 this week. 
    - The w12 query had x3=0.450 and produced the worst result in months. 
    - While x3=0.450 is technically at the boundary rather than over it, the pattern of poor results when x3 approaches 0.45 suggests the effective safe zone ends slightly below the hard ceiling. 
    
- Lowering to 0.440 adds a 0.010 safety buffer that all previous strong results comfortably satisfy: w9 best (x3=0.412), w11 best (x3=0.426), w10 result (x3=0.311) which are all well below 0.440.

```python
Y_transformed = power_transform(Y_updated.reshape(-1, 1), method='yeo-johnson').ravel()

print('Original Y:', Y_updated)
print('Transformed Y:', Y_transformed)

kernel = Matern(nu=2.5) + WhiteKernel(noise_level=1e-5)

model = GaussianProcessRegressor(
    kernel=kernel,
    n_restarts_optimizer=20)

model.fit(X_unique, Y_transformed)
```

## **Acquisition Function**

- EI with ξ=0.01 and std=0.020 centred on the w11 best [0.178, 0.332, 0.426, 0.199, 0.304, 0.745]. 
    - The ξ=0.01 is maintained from the strategy that produced the w11 breakthrough as it has proven to be the right level of exploitation pressure for this function. 
    
    - The std=0.020 is slightly tighter than the w11 std=0.025 that produced the crash
        - the w12 crash occurred because the acquisition function reached a point that was 0.420 units away from the centre in x1 (0.598 vs 0.178), which is 21 standard deviations at std=0.020. 
            - This means the crash point was selected from the uniform 15% component, not the focused component, suggesting the uniform component occasionally introduces catastrophic candidates. 
            
- For W13, the 85/15 split is maintained but the hard constraint on x3≤0.440 acts as a filter across both the focused and uniform candidates, ensuring no query from either component can enter the danger zone.

```python
def expected_improvement(X, model, y_max, xi=0.01):
    mu, sigma = model.predict(X, return_std=True)
    sigma = sigma.reshape(-1, 1); mu = mu.reshape(-1, 1)

    with np.errstate(divide='warn'):
        improvement = mu - y_max - xi
        Z = improvement / sigma
        ei = improvement * norm.cdf(Z) + sigma * norm.pdf(Z)
        ei[sigma == 0.0] = 0.0
    return ei

w_best = np.array([0.177775, 0.331618, 0.426355, 0.198583, 0.303865, 0.744797])

x_focused = np.random.normal(loc=w_best, scale=0.02, size=(8500, 6))
x_uniform = np.random.uniform(0, 1, size=(1500, 6))
x_grid = np.vstack([x_focused, x_uniform])
# Lower x3 ceiling to 0.440 — extra safety margin after W12 crash
x_grid[:, 2] = np.minimum(x_grid[:, 2], 0.440)
x_grid = np.clip(x_grid, 0, 1)

y_max = np.max(Y_transformed)
ei_values = expected_improvement(x_grid, model, y_max, xi=0.01)
next_query = x_grid[np.argmax(ei_values)]

print(f'Best coordinates: {w_best}')
print(f'Week 13 Query Recommendation (Function 7): {next_query[0]:.6f}-{next_query[1]:.6f}-{next_query[2]:.6f}-{next_query[3]:.6f}-{next_query[4]:.6f}-{next_query[5]:.6f}')
```
