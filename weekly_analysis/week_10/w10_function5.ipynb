## **Bayesian Optimisation** - Gaussian Process Regressor

- **Model:** Gaussian Process Regressor with Matern (nu=2.5) + WhiteKernel and Yeo-Johnson transformation.

- **Changes for W10 and Reasoning:**
    - I am keeping the same kernel 
        - The regression was caused by the acquisition function sampling a low-$x_2$ candidate, not by a model fit failure.
    - The W9 regression now provides the GP with a critical piece of information: 
        - [0.945, 0.489, 0.976, 0.982] is dramatically worse than [0.846, 1.000, 0.965, 1.000]. 
            - This contrast sharpens the GP's gradient estimate in $x_2$ specifically.
    - I am tightening the focused grid std to 0.008 (same as W9 intent but now re-centred correctly on the W8 best rather than last week's drifted version).

- **Theoretical Impact:** The stark W9 regression gives the GP model an extremely strong negative gradient signal in the $x_2$ dimension at values below 0.5. 
    - Combined with the W8 peak at $x_2=1.000$, the GP will now assign a very short ARD length scale to $x_2$, treating it as the most sensitive dimension. 
        - This is exactly what ARD is designed to learn and the contrast between 5896 and 3445 is large enough to make this gradient unmistakable.

```python
# Yeo-Johnson Output Transformation (HEBO)
# Wide output range (142 to 5896) requires this for stable GP calibration
Y_transformed = power_transform(Y_updated.reshape(-1, 1), method='yeo-johnson').ravel()

print('Original Y: ', Y_updated)
print('Transformed Y:', Y_transformed)

# Matern 2.5 — smooth kernel appropriate for the unimodal yield surface
kernel = Matern(nu=2.5) + WhiteKernel(noise_level=1e-5)
model = GaussianProcessRegressor(kernel=kernel, n_restarts_optimizer=20)
model.fit(X_unique, Y_transformed)
```

## **Acquisition Function** - Expected Improvement (EI) with Extreme Gradient Refinement

- **Strategy:** Expected Improvement (EI) with $\xi$ tightened to 0.0005.

- **Reasoning:** The function is confirmed unimodal. 
    - I know exactly where the peak is (w8 best) and I know $x_2$ must be near 1.000. 
        - Moving the $\xi$ to 0.0005 is appropriate because we are in the final climb toward the true peak.

- **Key Changes:**
    - $\xi$ tightened to 0.0005 (from 0.0005 last week so the same value, maintained).
    - Focused grid re-centred on the w8 all-time best [0.846, 1.000, 0.965, 1.000] with std=0.008.
    - 95/5 focused-to-uniform split as the w9 regression confirms broad exploration is counterproductive for this function.

- **Intended Impact:** Return to and surpass the w8 best of 5896.09 by keeping $x_2$ pinned near 1.000 and micro-tuning the remaining three dimensions.

```python
def expected_improvement(X, model, y_max, xi=0.0005):
    mu, sigma = model.predict(X, return_std=True)
    sigma = sigma.reshape(-1, 1)
    mu = mu.reshape(-1, 1)

    with np.errstate(divide='warn'):
        improvement = mu - y_max - xi
        Z = improvement / sigma
        ei = improvement * norm.cdf(Z) + sigma * norm.pdf(Z)
        ei[sigma == 0.0] = 0.0
    return ei

# Re-center on W8 all-time best
w8_best = np.array([0.845525, 0.999999, 0.964822, 0.999561])

# 95% Focused, 5% Uniform — extreme exploitation for a confirmed unimodal function
x_focused = np.random.normal(loc=w8_best, scale=0.008, size=(9500, 4))
x_uniform = np.random.uniform(0, 1, size=(500, 4))
x_grid = np.clip(np.vstack([x_focused, x_uniform]), 0, 1)

y_max = np.max(Y_transformed)
ei_values = expected_improvement(x_grid, model, y_max, xi=0.0005)
next_query = x_grid[np.argmax(ei_values)]

print(f'W8 best coordinates: {w8_best}')
print(f'Strategic Week 10 Query (Function 5): {next_query[0]:.6f}-{next_query[1]:.6f}-{next_query[2]:.6f}-{next_query[3]:.6f}')
```
