## **Bayesian Optimisation** - Gaussian Process Regressor

- **Model:** Gaussian Process Regressor with Matern (nu=2.5) + WhiteKernel and Yeo-Johnson transformation.

- **Changes for W10 and Reasoning:**
    - I am keeping the same kernel — the W9 breakthrough completely validates the model architecture.
    - I am re-centering the focused grid on the new W9 all-time best [0.097, 0.417, 0.412, 0.162, 0.308, 0.711] — not the W5 coordinates used in W9. 
        - The W9 result is the new anchor.
    
    - I am tightening the focused grid std to 0.025 (from 0.04 in W9). 
        - Now that we have a confirmed new peak, a tighter grid reduces the chance of wandering to sub-optimal $x_3$ values.
    
    - The $x_3 \leq 0.45$ hard ceiling is retained as we found out that it was the most important constraint for this function.

- **Theoretical Impact:** With a new all-time best confirmed, the GP's posterior will concentrate around [0.097, 0.417, 0.412, 0.162, 0.308, 0.711]. 
    - The ARD structure will now meaningfully learn that $x_3$ near 0.41 is optimal, the contrast between $x_3 = 0.412$ (2.134) and $x_3 = 0.588$ (0.838 in W8) gives the model a clear negative gradient in the $x_3$ direction above 0.45.

```python
# Yeo-Johnson Output Transformation (HEBO)
Y_transformed = power_transform(Y_updated.reshape(-1, 1), method='yeo-johnson').ravel()

print('Original Y:   ', Y_updated)
print('Transformed Y:', Y_transformed)

# Matern 2.5 — appropriate for the smooth performance score surface
# WhiteKernel accounts for stochastic variation in ML evaluation scores
kernel = Matern(nu=2.5) + WhiteKernel(noise_level=1e-4)
model = GaussianProcessRegressor(kernel=kernel, n_restarts_optimizer=20)
model.fit(X_unique, Y_transformed)
```

## **Acquisition Function**

- **Strategy:** Expected Improvement (EI) with $\xi = 0.01$, maintained from W9.

- **Reasoning:** EI anchored to the W9 best of 2.134 is the correct strategy.

    - The $x_3$ ceiling constraint ensures the acquisition function never recommends a point in the failure zone above 0.45.

- **Key Changes:**
    - Focused grid re-centred on the w9 all-time best [0.097, 0.417, 0.412, 0.162, 0.308, 0.711] with `std=0.025` (tighter than W9's 0.04).
    - $x_3$ hard ceiling at 0.45 maintained and applied to candidate grid.
    - 85/15 focused-to-uniform split.

- **Intended Impact:** Push the performance score above 2.134 by fine-tuning the six hyperparameters around the confirmed optimal configuration.

```python
def expected_improvement(X, model, y_max, xi=0.01):
    mu, sigma = model.predict(X, return_std=True)
    sigma = sigma.reshape(-1, 1)
    mu = mu.reshape(-1, 1)
    with np.errstate(divide='warn'):
        improvement = mu - y_max - xi
        Z = improvement / sigma
        ei = improvement * norm.cdf(Z) + sigma * norm.pdf(Z)
        ei[sigma == 0.0] = 0.0
    return ei

# Re-center on W9 new all-time best
w9_best = np.array([0.096872, 0.416732, 0.412049, 0.161576, 0.307971, 0.710714])

# 85% Focused, 15% Uniform
x_grid = np.random.normal(loc=w9_best, scale=0.025, size=(8500, 6))
x_uniform_part = np.random.uniform(0, 1, size=(1500, 6))
x_grid = np.vstack([x_grid, x_uniform_part])

# Hard ceiling: x3 must not exceed 0.45
x_grid[:, 2] = np.minimum(x_grid[:, 2], 0.45)
x_grid = np.clip(x_grid, 0, 1)

y_max = np.max(Y_transformed)
ei_values = expected_improvement(x_grid, model, y_max, xi=0.01)
next_query = x_grid[np.argmax(ei_values)]

print(f'W9 best coordinates: {w9_best}')
print(f'Strategic Week 10 Query (Function 7): {next_query[0]:.6f}-{next_query[1]:.6f}-{next_query[2]:.6f}-{next_query[3]:.6f}-{next_query[4]:.6f}-{next_query[5]:.6f}')
```
