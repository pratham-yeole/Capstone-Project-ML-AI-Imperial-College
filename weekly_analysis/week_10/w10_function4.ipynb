## **Bayesian Optimisation** - Gaussian Process Regressor

- **Model:** Gaussian Process Regressor with Matern (nu=2.5) + WhiteKernel and Yeo-Johnson transformation.

- **Changes for W10 and Reasoning:**
    - I am **switching from UCB to EI** for this function. 
        - Now that we have three positive results clearly clustered around [0.35–0.40], we have enough evidence to let EI drive the search, it will only recommend a new point if it has a meaningful probability of beating 0.602.
    - The Yeo-Johnson transformation is retained as the output range remains too wide for stable GP calibration without it.

- I am tightening the focused grid std to 0.010 (from 0.015 in w9) and shifting the centre to the w6 all-time best [0.403, 0.361, 0.354, 0.347].

- **Theoretical Impact:** The switch from UCB to EI fundamentally changes the selection criterion. 
    - Instead of rewarding uncertainty, EI requires the model to predict a high probability of improvement over 0.602.
        - This acts as a mathematical fence around the confirmed positive region and prevents the query from being pulled to adjacent unexplored areas with high variance.

```python
# Yeo-Johnson Output Transformation (HEBO)
Y_transformed = power_transform(Y_updated.reshape(-1, 1), method='yeo-johnson').ravel()

print('Original Y:   ', Y_updated)
print('Transformed Y:', Y_transformed)

# Matern 2.5 + WhiteKernel — maintained as it correctly maps the rugged landscape
kernel = Matern(nu=2.5) + WhiteKernel(noise_level=1e-5)
model = GaussianProcessRegressor(kernel=kernel, n_restarts_optimizer=20)
model.fit(X_unique, Y_transformed)
```

## **Acquisition Function** - Expected Improvement (EI)

- **Strategy:** Switching from UCB to **Expected Improvement (EI)** with $\xi = 0.05$.

- **Reasoning:** UCB has failed to keep queries in the confirmed positive zone for two consecutive weeks. 
    - The problem is that UCB rewards uncertainty regardless of likely improvement, so it keeps recommending unexplored dimensions of the 4D space that have not proven productive. 
    - EI is mathematically anchored to the current best of 0.602 and will not recommend a point unless it has a genuine predicted probability of beating that threshold.

- **Key Changes:**
    - Acquisition function switched from UCB ($\beta=3.0$) to EI ($\xi=0.05$).
    - Focused grid re-centred on the W6 all-time best [0.403, 0.361, 0.354, 0.347] with std=0.010.
    - 90/10 focused-to-uniform split.

- **Intended Impact:** Return to and surpass +0.602 by keeping the query anchored to the narrow confirmed positive ridge rather than drifting to uninformative high-uncertainty regions.

```python
def expected_improvement(X, model, y_max, xi=0.05):
    mu, sigma = model.predict(X, return_std=True)
    sigma = sigma.reshape(-1, 1)
    mu = mu.reshape(-1, 1)
    with np.errstate(divide='warn'):
        improvement = mu - y_max - xi
        Z = improvement / sigma
        ei = improvement * norm.cdf(Z) + sigma * norm.pdf(Z)
        ei[sigma == 0.0] = 0.0
    return ei

# Re-center on W6 all-time best
w6_best = np.array([0.402567, 0.361014, 0.353681, 0.347064])

# 90% Focused, 10% Uniform
x_focused = np.random.normal(loc=w6_best, scale=0.01, size=(9000, 4))
x_uniform = np.random.uniform(0, 1, size=(1000, 4))
x_grid = np.clip(np.vstack([x_focused, x_uniform]), 0, 1)

y_max = np.max(Y_transformed)
ei_values = expected_improvement(x_grid, model, y_max, xi=0.05)
next_query = x_grid[np.argmax(ei_values)]

print(f'W6 best coordinates: {w6_best}')
print(f'Strategic Week 10 Query (Function 4): 'f'{next_query[0]:.6f}-{next_query[1]:.6f}-{next_query[2]:.6f}-{next_query[3]:.6f}')
```
